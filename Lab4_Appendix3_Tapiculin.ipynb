```python
# %% Student inputs (shared)
from collections import defaultdict, Counter

LAST_NAME = "Tapiculin"
STUDENT_ID = "TUPM-26-1932"
SEED_NUM = int(STUDENT_ID[-1])
FAVORITE_ARTIST = "Shawn Mendes"
ART = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]   # character positions of the artist name
print(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 45.0, "PEN": 12.0, "USB": 350.0, "HEADPHONES": 800.0, "CALCULATOR": 600.0}

# 1. Generate transactions: quantity from SEED_NUM and artist character positions
transactions = []
for k in range(10):
    cust = customers[k % 5]
    prod = products[(k * 2 + 1 + k // 5) % 5]
    qty = (ord(ART[k]) * SEED_NUM) % 5 + 1
    transactions.append((cust, prod, qty))
transactions.append(transactions[0])                 # repeated customer-product purchase

print("=== 1. Generated Transaction Records ===")
for t in transactions: print(t)

# 2/3. Organize and compute values
records = [(c, p, q, round(q * prices[p], 2)) for c, p, q in transactions]
organized = defaultdict(lambda: defaultdict(list))
for c, p, q, v in records: organized[c][p].append((q, v))
print("\n=== 2. Organized Transaction Data ===")
for c, ps in organized.items(): print(c, dict(ps))
print("\n=== 4. Transaction Values ===")
for r in records: print(r)
overall = round(sum(r[3] for r in records), 2)

# 4. Repeated / distinct
pair_counts = Counter((c, p) for c, p, _, _ in records)
repeated = [p for p, n in pair_counts.items() if n > 1]
print("\n=== 3. Repeated/Distinct Transactions ===")
print("Repeated customer-product pairs:", repeated)
print("Distinct customers:", len({r[0] for r in records}), "| Distinct products:", len({r[1] for r in records}))

# 5. Summaries
cust_total, prod_total, prod_qty = defaultdict(float), defaultdict(float), defaultdict(int)
for c, p, q, v in records:
    cust_total[c] += v; prod_total[p] += v; prod_qty[p] += q
print("\n=== 5. Customer Summaries ===")
for c, v in cust_total.items(): print(c, round(v, 2))
print("\n=== 6. Product Summaries ===")
for p in prod_total: print(p, "qty", prod_qty[p], "value", round(prod_total[p], 2))

# 6. Ordered results
print("\n=== 7. Ordered Results (customers by total value) ===")
for rank, (c, v) in enumerate(sorted(cust_total.items(), key=lambda kv: kv[1], reverse=True), 1):
    print(rank, c, round(v, 2))

print("\n=== 8. Overall Transaction Summary / Final Consolidated Transaction Report ===")
print(f"Transactions: {len(records)} | Total value: {overall} | Total quantity: {sum(prod_qty.values())}")
print("Top product by value:", max(prod_total, key=prod_total.get))
```
