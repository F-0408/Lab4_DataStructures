```python
# %% Student inputs (shared)
from collections import defaultdict, Counter

LAST_NAME = "Tapiculin"
STUDENT_ID = "TUPM-26-1932"
SEED_NUM = int(STUDENT_ID[-1])
FAVORITE_ARTIST = "Shawn Mendes"
ART = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]   # character positions of the artist name
print(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

def category(score):
    if score >= 90: return "OUTSTANDING"
    if score >= 80: return "SATISFACTORY"
    if score >= 75: return "FAIR"
    return "NEEDS IMPROVEMENT"

# 1. Generate records using SEED_NUM and artist character positions
records = []
for i, st in enumerate(students):
    for j, sub in enumerate(subjects):
        score = 60 + (ord(ART[(i * 5 + j) % len(ART)]) * SEED_NUM + i + j) % 40
        records.append([st, sub, score])
records.append([students[0], "PROGRAMMING", 88])   # repeated student-subject record
for r in records:
    if r[0] == students[3] and r[1] == "ENGLISH":
        r[2] = None                                 # incomplete record (missing score)

print("=== 1. Generated Performance Records ===")
for r in records: print(r)

# 2. Organize student/subject data
organized = defaultdict(lambda: defaultdict(list))
for st, sub, sc in records:
    organized[st][sub].append(sc)
print("\n=== 2. Organized Student/Subject Data ===")
for st, subs in organized.items(): print(st, dict(subs))

# 3/4. Repeated + incomplete, categories
pair_counts = Counter((st, sub) for st, sub, _ in records)
repeated = [p for p, n in pair_counts.items() if n > 1]
incomplete = [(st, sub) for st, sub, sc in records if sc is None]
valid_scores = [(st, sub, sc) for st, sub, sc in records if sc is not None]
cats = [(st, sub, sc, category(sc)) for st, sub, sc in valid_scores]
print("\n=== 3. Repeated/Incomplete Records ===")
print("Repeated:", repeated)
print("Incomplete:", incomplete)
print("\n=== 4. Performance Categories ===")
print(Counter(c[3] for c in cats))

# 5. Student and subject summaries (valid scores only)
by_student, by_subject = defaultdict(list), defaultdict(list)
for st, sub, sc in valid_scores:
    by_student[st].append(sc); by_subject[sub].append(sc)
student_avg = {s: round(sum(v) / len(v), 2) for s, v in by_student.items()}
subject_avg = {s: round(sum(v) / len(v), 2) for s, v in by_subject.items()}
print("\n=== 5. Student and Subject Summaries ===")
print("Student averages:", student_avg)
print("Subject averages:", subject_avg)

# 6. Ordered results
print("\n=== 6. Ordered Results ===")
for rank, (s, a) in enumerate(sorted(student_avg.items(), key=lambda kv: kv[1], reverse=True), 1):
    print(rank, s, a, category(a))

print("\n=== Final Student Performance Report ===")
print(f"Records: {len(records)} | Valid: {len(valid_scores)} | Repeated: {len(repeated)} | Incomplete: {len(incomplete)}")
print("Top student:", max(student_avg, key=student_avg.get), "| Easiest subject:", max(subject_avg, key=subject_avg.get))
```
