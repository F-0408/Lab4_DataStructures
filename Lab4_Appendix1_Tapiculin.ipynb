```python
# %% Student inputs (shared)
from collections import defaultdict, Counter

LAST_NAME = "Tapiculin"
STUDENT_ID = "TUPM-26-1932"
SEED_NUM = int(STUDENT_ID[-1])
FAVORITE_ARTIST = "Shawn Mendes"
ART = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]   # character positions of the artist name
print(LAST_NAME, SEED_NUM, FAVORITE_ARTIST)


sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# 1. Generate records: (source, category, value) using SEED_NUM and artist character positions
records = []
for i, src in enumerate(sources):
    cat = categories[i]
    records.append((src, cat, round((ord(ART[i]) * SEED_NUM) % 100 + i / 10, 1)))
# extra measurements under two categories (multiple measurements)
for j in range(3):
    records.append((sources[0], "TEMPERATURE", round((ord(ART[5 + j]) + SEED_NUM * j) % 100, 1)))
    records.append((sources[1], "VOLTAGE", round((ord(ART[8 + j]) + SEED_NUM * j) % 100, 1)))
records.append(records[0])                       # one repeated test-source/category record

print("=== 1. Generated Test Data ===")
for r in records: print(r)

# 2. Organize by source and category
organized = defaultdict(lambda: defaultdict(list))
for src, cat, val in records:
    organized[src][cat].append(val)
print("\n=== 2. Organized Test Data ===")
for src, cats in organized.items(): print(src, dict(cats))

# 3. Repeated and distinct data
counts = Counter(records)
repeated = [r for r, n in counts.items() if n > 1]
distinct_records = set(records)
print("\n=== 3. Repeated/Distinct Data ===")
print("Repeated records:", repeated)
print("Distinct records:", len(distinct_records), "of", len(records))
print("Distinct sources:", len({r[0] for r in records}), "| Distinct categories:", len({r[1] for r in records}))

# 4. Numerical and categorical summaries
by_cat = defaultdict(list)
for _, cat, val in records: by_cat[cat].append(val)
summary = {c: {"count": len(v), "min": min(v), "max": max(v), "avg": round(sum(v) / len(v), 2)} for c, v in by_cat.items()}
cat_counts = Counter(r[1] for r in records)
print("\n=== 4. Numerical and Categorical Summaries ===")
for c, s in summary.items(): print(c, s)
print("Records per category:", dict(cat_counts))

# 5/6. Processed + ordered results (highest average first)
ordered = sorted(summary.items(), key=lambda kv: kv[1]["avg"], reverse=True)
print("\n=== 5. Processed Results ===")
print({c: s["avg"] for c, s in summary.items()})
print("\n=== 6. Ordered Results (by average, descending) ===")
for rank, (c, s) in enumerate(ordered, 1): print(rank, c, s["avg"])

# 7. Final summary
print("\n=== 7. Final Summary / Final Output ===")
print(f"{len(records)} records | {len(distinct_records)} distinct | {len(repeated)} repeated | "
      f"highest average: {ordered[0][0]} ({ordered[0][1]['avg']})")
```
